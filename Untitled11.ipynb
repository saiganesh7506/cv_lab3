```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
rmin, rmax = 100, 200
img = cv2.imread("/content/balaram.jpg", 0)
slicing_wo_background = np.zeros(img.shape, dtype=np.uint8)
slicing_w_background = img.copy()

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        pixel_value = img[i][j]
        if rmin <= pixel_value <= rmax:
            slicing_wo_background[i][j] = 255
            slicing_w_background[i][j] = 255

def show(*images):
    fig, axes = plt.subplots(1, len(images), figsize=(15, 5))
    if len(images) == 1:
        axes.imshow(images[0], cmap='gray')
        axes.axis('off')
    else:
        for ax, img_to_show in zip(axes, images):
            ax.imshow(img_to_show, cmap='gray')
            ax.axis('off')
    plt.show()

show(img, slicing_wo_background, slicing_w_background)
```
